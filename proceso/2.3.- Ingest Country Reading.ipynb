```python
# Celda 1
dbutils.widgets.removeAll()
```

```python
# Celda 2
from pyspark.sql.functions import current_timestamp
```

```python

# Celda 3
dbutils.widgets.text("catalogo", "catalog_au")
dbutils.widgets.text("esquema", "bronze")
dbutils.widgets.text("sqlServerName", "ssqlaizaproyect")
dbutils.widgets.text("sqlDatabaseName", "asqlaizaproyecto")
```

```python
# Celda 4
catalogo = dbutils.widgets.get("catalogo")
esquema = dbutils.widgets.get("esquema")
sqlServerName = dbutils.widgets.get("sqlServerName")
sqlDatabaseName = dbutils.widgets.get("sqlDatabaseName")

sql_user = dbutils.secrets.get(scope="kv-scope", key="sql-admin-user")
sql_password = dbutils.secrets.get(scope="kv-scope", key="sql-admin-password")

jdbc_url = f"jdbc:sqlserver://{sqlServerName}.database.windows.net:1433;database={sqlDatabaseName};encrypt=true;trustServerCertificate=false;loginTimeout=30;"
```

```python
# Celda 5
df_country = spark.read \
    .format("jdbc") \
    .option("url", jdbc_url) \
    .option("dbtable", "dbo.country_reading") \
    .option("user", sql_user) \
    .option("password", sql_password) \
    .load()
```

```python
# Celda 6
country_final_df = df_country.withColumn("ingestion_date", current_timestamp())
```

```python
# Celda 7
country_final_df.write.mode("overwrite").insertInto(f"{catalogo}.{esquema}.country_reading")
```
