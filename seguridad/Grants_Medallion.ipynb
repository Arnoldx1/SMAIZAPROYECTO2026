## Grants

### Permisos a nivel de Catalogo

```python
%sql
GRANT USE CATALOG ON CATALOG catalog_au TO `Administradores`;
GRANT USE CATALOG ON CATALOG catalog_au TO `Desarrolladores`;
GRANT USE CATALOG ON CATALOG catalog_au TO `Analistas`;
GRANT USE CATALOG ON CATALOG catalog_au TO `Lectores`;
GRANT USE CATALOG ON CATALOG catalog_au TO `Auditores`;

GRANT ALL PRIVILEGES ON CATALOG catalog_au TO `Administradores`;
```

### Permisos Bronze (solo Admin y Desarrolladores)

```python
%sql
GRANT USE SCHEMA ON SCHEMA catalog_au.bronze TO `Desarrolladores`;
GRANT CREATE TABLE ON SCHEMA catalog_au.bronze TO `Desarrolladores`;
GRANT MODIFY ON SCHEMA catalog_au.bronze TO `Desarrolladores`;
GRANT SELECT ON SCHEMA catalog_au.bronze TO `Desarrolladores`;

GRANT USE SCHEMA ON SCHEMA catalog_au.bronze TO `Auditores`;
GRANT SELECT ON SCHEMA catalog_au.bronze TO `Auditores`;
```

### Permisos Silver (Desarrolladores + Analistas)

```python
%sql
GRANT USE SCHEMA ON SCHEMA catalog_au.silver TO `Desarrolladores`;
GRANT CREATE TABLE ON SCHEMA catalog_au.silver TO `Desarrolladores`;
GRANT MODIFY ON SCHEMA catalog_au.silver TO `Desarrolladores`;
GRANT SELECT ON SCHEMA catalog_au.silver TO `Desarrolladores`;

GRANT USE SCHEMA ON SCHEMA catalog_au.silver TO `Analistas`;
GRANT SELECT ON SCHEMA catalog_au.silver TO `Analistas`;

GRANT USE SCHEMA ON SCHEMA catalog_au.silver TO `Auditores`;
GRANT SELECT ON SCHEMA catalog_au.silver TO `Auditores`;
```

### Permisos Golden (todos leen, solo Dev/Admin escriben)

```python
%sql
GRANT USE SCHEMA ON SCHEMA catalog_au.golden TO `Desarrolladores`;
GRANT CREATE TABLE ON SCHEMA catalog_au.golden TO `Desarrolladores`;
GRANT MODIFY ON SCHEMA catalog_au.golden TO `Desarrolladores`;
GRANT SELECT ON SCHEMA catalog_au.golden TO `Desarrolladores`;

GRANT USE SCHEMA ON SCHEMA catalog_au.golden TO `Analistas`;
GRANT SELECT ON SCHEMA catalog_au.golden TO `Analistas`;

GRANT USE SCHEMA ON SCHEMA catalog_au.golden TO `Lectores`;
GRANT SELECT ON SCHEMA catalog_au.golden TO `Lectores`;

GRANT USE SCHEMA ON SCHEMA catalog_au.golden TO `Auditores`;
GRANT SELECT ON SCHEMA catalog_au.golden TO `Auditores`;
```
