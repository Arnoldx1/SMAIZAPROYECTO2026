```python
%sql
-- Crear el share
CREATE SHARE IF NOT EXISTS libreria_golden_share
COMMENT 'Share de las tablas Golden para consumo en Power BI';

-- Agregar las 5 tablas Golden al share
ALTER SHARE libreria_golden_share ADD TABLE catalog_au.golden.top_authors;
ALTER SHARE libreria_golden_share ADD TABLE catalog_au.golden.top_books;
ALTER SHARE libreria_golden_share ADD TABLE catalog_au.golden.genre_distribution;
ALTER SHARE libreria_golden_share ADD TABLE catalog_au.golden.top_books_by_genre;
ALTER SHARE libreria_golden_share ADD TABLE catalog_au.golden.country_reading;

-- Verificar que se agregaron bien
SHOW ALL IN SHARE libreria_golden_share;
```

```python
%sql
-- Crear un recipient abierto (no-Databricks), para usar el conector de Power BI
CREATE RECIPIENT IF NOT EXISTS power_bi_recipient;

-- Darle permiso de lectura sobre el share
GRANT SELECT ON SHARE libreria_golden_share TO RECIPIENT power_bi_recipient;

-- Obtener el link de activacion (para descargar el archivo de credenciales)
DESCRIBE RECIPIENT power_bi_recipient;
```
